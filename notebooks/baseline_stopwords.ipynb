# Baseline Stopword Ratio Experiments

## Import

```python
from pathlib import Path
import sys

import pandas as pd
from IPython.display import display
from omegaconf import OmegaConf

PROJECT_ROOT = Path.cwd().parent
sys.path.insert(0, str(PROJECT_ROOT / "src"))

from pipelines.pipeline import train_pipeline

pd.set_option("display.max_columns", None)
pd.set_option("display.precision", 4)

config = OmegaConf.load(PROJECT_ROOT / "src" / "configs" / "experiment.yaml")
print(OmegaConf.to_yaml(config))
```

## Run Experiments

```python
metrics = train_pipeline(
    categories=list(config.categories),
    stopword_ratios=list(config.stopword_ratios),
    model_names=list(config.model_names),
    data_dir=PROJECT_ROOT / config.paths.data_dir,
    model_path=PROJECT_ROOT / config.paths.model_path,
    output_dir=PROJECT_ROOT / config.paths.results_dir,
    images_dir=PROJECT_ROOT / config.paths.images_dir,
)

metrics = metrics.sort_values(["stopword_ratio", "model"]).reset_index(drop=True)
metric_columns = [
    "stopword_ratio",
    "model",
    "accuracy",
    "macro_f1",
    "weighted_f1",
    "cv_best_score",
    "best_params",
]

display(metrics[metric_columns])
```

## Baseline View: No Stopword Removal

```python
baseline_metrics = (
    metrics[metrics["stopword_ratio"] == 0.0]
    .sort_values("macro_f1", ascending=False)
    .reset_index(drop=True)
)

baseline_columns = [
    "model",
    "accuracy",
    "macro_f1",
    "weighted_f1",
    "cv_best_score",
    "best_params",
]

display(baseline_metrics[baseline_columns])

best_baseline = baseline_metrics.iloc[0]
print(
    "Best baseline model: "
    f"{best_baseline['model']} | "
    f"accuracy={best_baseline['accuracy']:.4f} | "
    f"macro_f1={best_baseline['macro_f1']:.4f} | "
    f"best_params={best_baseline['best_params']}"
)
```
